**CLEANING AND ORGANIZING THE DATA-**

```python
from google.colab import files

uploaded = files.upload()
```

```python
import pandas as pd

df = pd.read_csv("students_missing_duplicates.csv")
print(df)
```

Check missing values-

```python
print(df.isnull().sum())
```

Check duplicate records-


```python
print(df.duplicated().sum())
```

Remove duplicate records-

```python
df = df.drop_duplicates()
print(df)
```

Fill missing numerical values-

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df["Attendance_Percent"] = df["Attendance_Percent"].fillna(df["Attendance_Percent"].mean())

print(df)
```

Fill missing text values-

```python
df["Name"] = df["Name"].fillna("Unknown")
df["Branch"] = df["Branch"].fillna("Not Available")

print(df)
```

Check missing values again-

```python
print(df.isnull().sum())
```

**Basic Data Analysis-**

Data Analysis-

Data analysis is the process of examining data to find useful information and understand patterns.

Basic operations in Data Analysis--
1. Mean – Finds the average value.
2. Median – Finds the middle value.
3. Maximum – Finds the highest value.
4. Minimum – Finds the lowest value.
5. Count – Counts the number of records.
6. Grouping – Groups data based on a column.

Practical programs-


```python
print(df.describe())
```

```python
print("Average Marks:", df["Marks"].mean())
```

```python
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())
```

```python
print("Total Students:", df["Name"].count())
```

```python
print(df.groupby("Branch")["Marks"].mean())
```

```python
print("Median Marks:", df["Marks"].median())
```

 I learned how to perform basic data analysis using Pandas. I calculated average, maximum and minimum marks, counted records and compared average marks branch-wise.
